## 2. List Comprehension vs For Loop

```python
numbers = list(range(1000))
squares = [n * n for n in numbers]
print(squares[:10])
```

```python
numbers = list(range(1000))
```

```python
%timeit [n * n for n in numbers]
```

```python
def square_loop(numbers):
    result = []
    for n in numbers:
        result.append(n * n)
    return result
```

```python
%timeit square_loop(numbers)
```

## 3. Execution Time Comparison

- List comprehension creates a list in a concise way.
- For loop creates the same result using iteration and append.
- %timeit measures the execution time of both methods.
- The method with the lower execution time is faster in this test.
